```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

def showImg(axes, img):
    return axes.imshow(img)
def showScaleImg(axes, img, scale):
    return axes.imshow(img,cmap=scale)
fig,axs = plt.subplots(1,3,figsize=(20,10))

img1 = cv2.imread("imagen-1.jpg", cv2.IMREAD_COLOR)
img2 = cv2.imread("imagen-2.jpg", cv2.IMREAD_COLOR)
img3 = cv2.imread("imagen-3.jpg", cv2.IMREAD_COLOR)
imgrgb1 = cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
imgrgb2 = cv2.cvtColor(img2,cv2.COLOR_BGR2RGB)
imgrgb3 = cv2.cvtColor(img3,cv2.COLOR_BGR2RGB)

showImg(axs[0],imgrgb1)
showImg(axs[1],imgrgb2)
showImg(axs[2],imgrgb3)
```

```python
fig,axs = plt.subplots(1,3,figsize=(20,10))
showScaleImg(axs[0],img1[...,2], "gray")
showScaleImg(axs[1],img2[...,2], "gray")
showScaleImg(axs[2],img3[...,2], "gray")

cv2.imwrite("imagen-1-grays.png",imgrgb1[...,0])
cv2.imwrite("imagen-2-grays.png",imgrgb2[...,0])
cv2.imwrite("imagen-3-grays.png",imgrgb3[...,0])
```

```python
def imprimirCanales(img, imgrgb, title):
    fig,axs = plt.subplots(2,4,figsize=(20,10))
    
    # Canales falsos
    showImg(axs[0][0],imgrgb)
    axs[0][0].set_title(title)

    showScaleImg(axs[0][1],imgrgb[...,0], "Reds")
    axs[0][1].set_title("Canal rojo")
    showScaleImg(axs[0][2],imgrgb[...,0], "Greens")
    axs[0][2].set_title("Canal verde")
    showScaleImg(axs[0][3],imgrgb[...,0], "Blues")
    axs[0][3].set_title("Canal azul")

    # Canales reales
    r = img[:,:,2]
    r_img = np.zeros_like(imgrgb)
    r_img[:,:,0] = r

    g = img[:,:,1]
    g_img = np.zeros_like(imgrgb)
    g_img[:,:,1] = g

    b = img[:,:,0]
    b_img = np.zeros_like(imgrgb)
    b_img[:,:,2] = b

    showImg(axs[1][0],imgrgb)
    axs[1][0].set_title(title)

    showImg(axs[1][1],r_img)
    axs[1][1].set_title("Canal rojo")
    showImg(axs[1][2],g_img)
    axs[1][2].set_title("Canal verde")
    showImg(axs[1][3],b_img)
    axs[1][3].set_title("Canal azul")
```

```python

imprimirCanales(img1, imgrgb1, "Imagen 1")
```

```python
imprimirCanales(img2, imgrgb2, "Imagen 2")
```

```python
imprimirCanales(img3, imgrgb3, "Imagen 3")
```
