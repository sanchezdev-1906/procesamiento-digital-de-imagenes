## Ejercicio 1

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

rimg = cv2.imread("imagen1.jpg")
imgrgb = cv2.cvtColor(rimg, cv2.COLOR_BGR2RGB)

x = 600
y = 600
width = 200
height = 200

recorte = imgrgb[y:y+height, x:x+width]

fig,axs = plt.subplots(1,2,figsize=(5,5))
axs[0].imshow(imgrgb)
axs[1].imshow(recorte)

recorte_res = cv2.cvtColor(recorte,cv2.COLOR_RGB2BGR)
cv2.imwrite("01-recorte.png",recorte_res)
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

rimg = cv2.imread("imagen1.jpg",cv2.IMREAD_GRAYSCALE)

mask = np.zeros((250,250),dtype=np.uint8)
rlen= len(mask)

for r in range(rlen):
    mask[r,:] = int(255*(r/rlen))


fig,axs = plt.subplots(1,1,figsize=(5,5))
axs.imshow(mask)

plt.show()
```

## Ejercicio 3

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = np.zeros(shape=(800,800))

img_size = 800
R = np.ones((img_size, img_size)) * 255
G = np.ones((img_size, img_size)) * 255
B = np.ones((img_size, img_size)) * 255


img = np.dstack((R,G,B))
img = np.array(img,dtype=int)
mask = np.zeros(img.shape[:2],dtype=np.uint8)



for r in range(8):
    for c in range(8):
        can = (r%2==0 and c%2==0) or (r%2==1 and c%2==1)
        if(can):
            x = int(r*100)
            y = int(c*100)
            cv2.rectangle(mask,(x,y),(x+100,y+100),255,-1)

result = cv2.bitwise_and(img,img,mask=mask)
fig,axs = plt.subplots(1,1,figsize=(5,5))
axs.imshow(result)

```
