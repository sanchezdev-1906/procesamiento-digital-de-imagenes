```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imgo = cv2.imread("imagen-1.jpg")
img = cv2.cvtColor(imgo,cv2.COLOR_BGR2RGB)

mask1 = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
mask2 = np.ones((25,25),np.float32)/(25*25)
mask3 = np.array([[0,0,0],[-1,1,0],[0,0,0]])*11
mask4 = np.array([[-2,-1,0],[-1,1,1],[0,1,2]])
mask5 = np.array([[0,1,0],[1,-4,1],[0,1,0]])*11
mask6 = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
mask7 = np.array([[1,-2,1],[-2,5,-2],[1,-2,1]])*5
mask8 = np.array([[1,1,1],[1,-2,1],[-1,-1,-1]])*5
mask9 = np.array([[-1,1,1],[-1,-2,1],[-1,1,1]])*5
mask10 = np.array([[1,2,3,1,1],[2,7,11,7,2],[3,11,17,11,3],[2,7,11,7,2],[1,2,3,1,1]])/81

final1 = cv2.filter2D(img, -1, mask1)
final2 = cv2.filter2D(img, -1, mask2)
final3 = cv2.filter2D(img, -1, mask3)
final4 = cv2.filter2D(img, -1, mask4)
final5 = cv2.filter2D(img, -1, mask5)
final6 = cv2.filter2D(img, -1, mask6)
final7 = cv2.filter2D(img, -1, mask7)
final8 = cv2.filter2D(img, -1, mask8)
final9 = cv2.filter2D(img, -1, mask9)
final10 = cv2.filter2D(img, -1, mask10)

fig,axs = plt.subplots(5,2,figsize=(10,15))
axs[0][0].imshow(final1)
axs[0][0].set_title("Enfoque")
axs[0][1].imshow(final2)
axs[0][1].set_title("Desenfoque")
axs[1][0].imshow(final3)
axs[1][0].set_title("Realce de bordes")
axs[1][1].imshow(final4)
axs[1][1].set_title("Repujado")
axs[2][0].imshow(final5)
axs[2][0].set_title("Deteccion de bordes")
axs[2][1].imshow(final6)
axs[2][1].set_title("Filtro de tipo Sobel")
axs[3][0].imshow(final7)
axs[3][0].set_title("Filtro de tipo Sharpen")
axs[3][1].imshow(final8)
axs[3][1].set_title("Filtro Norte")
axs[4][0].imshow(final9)
axs[4][0].set_title("Filtro Este")
axs[4][1].imshow(final10)
axs[4][1].set_title("Filtro de tipo Gauss")
```
