```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

img_bg = cv2.imread("imgprueba.jpg")
img_rgb = cv2.cvtColor(img_bg, cv2.COLOR_BGR2RGB)
print(f'Alto: {img_bg.shape[0]}')
print(f'Ancho: {img_bg.shape[1]}')
print(f'Canales: {img_bg.shape[2]}')
fig,axs = plt.subplots(1,4,figsize=(20,5))
axs[0].imshow(img_rgb)
axs[0].set_title("Imagen original")

fig_img = axs[1].imshow(img_rgb[...,0], cmap = "Reds")
fig.colorbar(fig_img, ax=axs[1],shrink = 0.5)
axs[1].set_title("Canal rojo")
fig_img = axs[2].imshow(img_rgb[...,0], cmap = "Greens")
fig.colorbar(fig_img, ax=axs[2],shrink = 0.5)
axs[2].set_title("Canal verde")
fig_img = axs[3].imshow(img_rgb[...,0], cmap = "Blues")
fig.colorbar(fig_img, ax=axs[3],shrink = 0.5)
axs[3].set_title("Canal azul")
```

```python
recorte = img_rgb[200:700,0:300,]
plt.imshow(recorte)
cv2.imwrite("r-recorte.png",cv2.cvtColor(recorte,cv2.COLOR_BGR2RGB))
```
