## Ejercicio 1
Desarrolle una aplicación que calcule el factorial de un numero entero introducido por teclado.

Entrada: 
```sh
6
```

````python
def factorial(n):
    acum = 1
    for i in range(2, n+1):
        acum *= i
    return acum

n = int(input("Ingrese un numero para calcular su factorial: "))
print("El factorial del numero es: " + str(factorial(n)))
````

## Ejercicio 2
 Desarrolle una aplicación que determine si un numero introducido por teclado es primo o
no.

Entrada:
```sh
131
```

````python
def esPrimo(num):
    num = int(num)
    if(num <= 1): 
        return False
    if(num == 2 or num == 3 ):
        return True
    if(num % 2 == 0 or num % 3 == 0):
        return False
    q = int(num**(1/2))
    for i in range(5, q + 1, 6):
        if num % i == 0: 
            return False
    return True

n = int(input("Ingrese un numero: "))
print(f"{n} {"es" if esPrimo(n) else "no es"} primo")
````

## Ejercicio 3
Desarrolle una aplicación que permita ingresar diferentes números por teclado con un limite
definido inicialmente (no menor a 3 números) y determine cual de ellos es el mayor y el menor.

Entrada:
```sh
5
4
20
3
-2
-50
```

````python
n = None

while (n == None):
    n = int(input("Ingrese un numero limite mayor o igual a 3.\n"))

menor = None

for i in range(n):
    num = int(input().strip())
    if(menor == None):
        menor = num
    if(num < menor):
        menor = num
print(f"El numero menor es {menor}")
````

## Ejercicio 4
Desarrolle una aplicación que permita identificar si una cadena introducida por teclado es
un palíndromo ( oso es palíndromo )

````python
s = input().strip()
slen = len(s)
palin = True
for i in range(slen):
    if s[i] != s[slen-1-i]:
        palin = False
print(f"\"{s}\" { "es"if palin else "no es"} palindromo")
````

## Ejercicio 5
Desarrolle una aplicación que permita introducir 7 números por teclado y determine cual de
ellos se repite más veces y cuantas.

Entrada:
```sh
7
6
5
7
6
7
5
```

````python
numbers = {}
mayor = {
    "rep": 0,
    "val": None
}
for i in range(7):
    k = input()
    n = int(k)
        
    if numbers.get(k) == None:
        numbers[k] = 1
    else:
        numbers[k] += 1

    if (mayor["val"] == None or numbers[k] > mayor["rep"]):
        mayor["rep"] = numbers[k]
        mayor["val"] = n
print(f"{mayor["val"]} se repite {mayor["rep"]} veces")
````
